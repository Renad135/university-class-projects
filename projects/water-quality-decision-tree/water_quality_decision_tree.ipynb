# Water Quality Classification with a Decision Tree

A class project that predicts a water sample safety label from chemical and biological measurements using a decision tree.

**Data required:** put `waterQuality1.csv` next to this notebook before running it. The dataset is not included.

The saved Colab run reported 95.83% accuracy and confusion matrix `[[2084, 35], [65, 215]]`. Because the split was not seeded, results can vary. For class 1, recall in that run was 76.8%. This coursework model is not validated for real water-safety decisions.


```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset (provide waterQuality1.csv in the working directory)
data = pd.read_csv('waterQuality1.csv')
data.columns
```

```python
# Select the 20 input features and the safety label
cols = ['aluminium', 'ammonia', 'arsenic', 'barium', 'cadmium', 'chloramine',
        'chromium', 'copper', 'flouride', 'bacteria', 'viruses', 'lead',
        'nitrates', 'nitrites', 'mercury', 'perchlorate', 'radium', 'selenium',
        'silver', 'uranium']
x = data[cols].values
y = data['is_safe'].values
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.3)
```

```python
# Train a decision tree with entropy as the split criterion
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(criterion='entropy', max_depth=6, min_samples_split=45)
dt.fit(X_train, y_train)
```

```python
# Display the learned tree rules
from sklearn.tree import export_text
print(export_text(dt, feature_names=cols))
```

```python
# Predict and evaluate on the held-out test data
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = dt.predict(X_test)
print('Confusion matrix:')
print(confusion_matrix(y_test, y_pred))
print(f'Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%')
```
